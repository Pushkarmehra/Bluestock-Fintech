# 📊 Indian Mutual Fund Industry — EDA Key Findings (2022–2025)

**Dataset:** AMFI India Mutual Fund Data | **Period:** Jan 2022 – Mar 2025  
**Scope:** AUM growth, SIP inflows, investor demographics, NAV trends, sector allocation, folio counts

---

## 🔍 Finding 1 — Explosive SIP Inflow Growth

- Monthly SIP inflows **nearly tripled** from ₹11,000 Cr (Jan 2022) to a record **₹31,002 Cr** milestone by early 2026.
- Growth has been **remarkably consistent** with no major sustained dips — even through the 2024 market correction phase.
- This reflects deep structural shift: Indian retail investors are committing to **disciplined, habit-based investing** via SIPs rather than lump-sum bets.

## 🔍 Finding 2 — Folio Count Doubled in ~4 Years

- Industry folio count grew from **13.26 Cr (2022)** to **26.12 Cr (early 2026)** — a near **2× increase** in under 4 years.
- Growth accelerated sharply from mid-2023 onward, suggesting a **post-COVID participation surge** and increased financial awareness among retail investors.
- The near-vertical trajectory in 2025–2026 signals **new investor onboarding** continuing at an accelerating pace.

## 🔍 Finding 3 — 26–35 Age Group Dominates Investor Base

- The **26–35 age cohort** comprises **41.1%** of all investors — the single largest segment by a wide margin.
- The **36–45 group** adds another 24.9%, meaning the **millennial demographic (26–45) accounts for ~66%** of the total investor base.
- Younger investors (18–25) represent 15% — a significant and growing cohort, likely driven by digital-first platforms and social media financial literacy.

## 🔍 Finding 4 — SBI Mutual Fund Leads AUM with ₹12.5 Lakh Crore

- **SBI Mutual Fund** dominates with AUM of **₹12.5 Lakh Crore** in 2025 — significantly ahead of all peers.
- **ICICI Prudential MF** (~₹10 L Cr) and **HDFC Mutual Fund** (~₹8.5 L Cr) are the next largest.
- All top-10 fund houses showed **consistent year-on-year AUM expansion** from 2022–2025, reflecting broad industry tailwinds rather than winner-take-all dynamics.

## 🔍 Finding 5 — Liquid Funds Dominate Category-Wise Net Inflows

- The **Liquid category** stands out starkly in the inflow heatmap — consistently recording the **highest net inflows** (orange/peach tones) across all months from Apr 2024 to Mar 2025.
- All other categories (Equity, ELSS, Hybrid, Mid/Small Cap, Sectoral) show **comparatively muted inflows** — the difference is an order of magnitude.
- This suggests **institutional and treasury participants** heavily use liquid funds for short-term parking, skewing category-level net inflow data.

## 🔍 Finding 6 — Banking & IT Lead Equity Portfolio Sector Allocation

- Across equity fund portfolios, **Banking (22.0%)** and **IT (15.3%)** are the top two sector allocations — together forming **over 37%** of aggregate equity AUM.
- **Pharma (13.7%)** and **Automobile (10.9%)** round out the top four, reflecting India's export-oriented and consumption-driven growth themes.
- Defensive sectors like **Utilities (8.9%)** and **FMCG (7.7%)** together account for ~17%, indicating a **moderate risk tilt** in fund portfolio construction.

## 🔍 Finding 7 — T30 Cities Contribute 66% of Investor Base

- **T30 (Top 30) cities** account for **66.3%** of investor distribution versus **33.7% from B30** (Beyond Top 30) cities.
- While urban concentration is expected, the **B30 share at 1-in-3 investors** indicates meaningful financial inclusion progress in smaller cities and towns.
- AMFI and SEBI's B30 incentive programs appear to be bearing fruit, though the **urban-rural gap remains substantial**.

## 🔍 Finding 8 — SIP Amount Distribution is Uniform Across Age Groups

- The SIP amount boxplot (log scale) reveals **strikingly similar distribution shapes** across all age groups (18–25 through 56+).
- Median SIP amounts cluster around **₹10,000–₹20,000** for all cohorts, with interquartile ranges roughly spanning ₹3,000–₹1,50,000.
- This uniformity suggests SIP ticket sizes are driven more by **platform defaults and product design** than by age or life-stage — the ₹500–₹1,000 minimum SIP has democratised participation.

## 🔍 Finding 9 — NAV Trends Show Low Cross-Fund Correlation

- The correlation matrix of NAV returns across 10 selected funds shows **predominantly near-zero off-diagonal correlations** (dark cells), with only a few fund pairs showing moderate positive correlation.
- This indicates that **selected funds track different benchmarks or strategies**, offering genuine diversification benefit when combined in a portfolio.
- Fund **100025** (red line in NAV trend chart) stands out as a clear **high NAV outperformer** (₹3,000–₹4,200+ range), while most others cluster below ₹1,000 NAV.

## 🔍 Finding 10 — State-Level SIP Distribution is Surprisingly Even

- The horizontal bar chart of SIP amounts by state reveals a **narrow band of variation** — total SIP amounts range from ~₹27 Cr (Maharashtra) to ~₹32 Cr (Punjab) across the top 12 states.
- Notably, **Punjab and Tamil Nadu lead** over traditional financial hubs like Maharashtra and Karnataka — possibly reflecting **higher SIP penetration rates** in semi-urban populations relative to total investor count.
- The **evenness of the distribution** (all states within ~15% of each other) suggests SIP adoption is becoming a **pan-India phenomenon** rather than metro-concentrated.

---

> **Summary:** The Indian MF industry between 2022–2025 shows hallmarks of a **structurally maturing market**: rising SIP discipline, young investor dominance, accelerating folio growth, geographic diversification, and category breadth — all pointing to sustained long-term growth.
